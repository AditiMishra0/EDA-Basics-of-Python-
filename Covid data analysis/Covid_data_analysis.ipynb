```python
import numpy as np
import pandas as pd
```

```python
#Importing dataset

data = pd.read_csv("/content/covid_19_data.csv")
data.head()
```

```python
data = data.drop(['SNo','Last Update'],axis=1)
data.head()
```

```python
data.info()
```

```python
data['ObservationDate'] = data['ObservationDate'].astype('datetime64[ns]')
data['Confirmed'] = data['Confirmed'].fillna(0).astype('int64')
data['Deaths'] = data['Deaths'].fillna(0).astype('int64')
data['Recovered'] = data['Recovered'].fillna(0).astype('int64')
data.info()
```

```python
#Active cases

data['Active'] = data['Confirmed'] - data['Recovered'] - data['Deaths']
data.head(25)

data.iloc[60:77]
```

```python
#Last/Current date

data['ObservationDate'].max()
```

```python
#Fetch the latest date data

latest_data = data[data['ObservationDate'] == data['ObservationDate'].max()]
latest_data.head()
```

```python
latest_data.shape
```

```python
latest_data['Country/Region'].value_counts()
```

```python
latest_data['Country/Region'].unique()
```

```python
latest_data['Country/Region'].nunique()
```

```python
countries = latest_data.groupby('Country/Region')[["Confirmed", "Deaths", "Recovered", "Active"]].sum()
countries = countries.reset_index()
countries.head()
```

```python
countries[countries['Country/Region'] == "India"]
```

##Plotting on the World map

```python
import plotly.express as px
```

```python
world_map = px.choropleth(countries,locations = "Country/Region", locationmode = "country names", color= "Confirmed", color_continuous_scale="fall", range_color=[0,1000000])
world_map
```

##Top 20 countries with the maximum confirmed and recovered cases

```python
from pandas.core import groupby

top = latest_data.groupby('Country/Region')[['Confirmed', 'Recovered']]. sum().reset_index()
top = top.sort_values(['Confirmed'], ascending=False)

top_20 = top.head(20)
top_20
```

##Analysis on India

```python
india = data[data['Country/Region']=="India"]
india
```

```python
#Total number of provinces and states

india['Province/State'].nunique()
```

```python
#List of province and states

india['Province/State'].unique()
```

```python
#Fill na with mode

india['Province/State'] = india['Province/State'].fillna(india['Province/State'].mode())
india
```

```python
#Fill unknown with mode

india['Province/State'] = india['Province/State'].replace('Unknown', india['Province/State'].mode()[0])
india
```

```python
india_latest_date = india[india['ObservationDate']==india['ObservationDate'].max()]
india_latest_date
```

```python
top_state = india_latest_date.groupby("Province/State")[['Confirmed', 'Recovered']].sum().reset_index()
top_state = top_state.sort_values(['Confirmed'], ascending=False)
top_state
```

```python
top_state['Confirmed'].max()
```
